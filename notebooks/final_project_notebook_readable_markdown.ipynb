<h3 style="color:#073B74;">۳. از داده‌ی خام تا داده‌ی قابل‌تحلیل</h3>
<p>داده خام سه پلتفرم ساختار یکسانی ندارد؛ بنابراین قبل از تحلیل باید به قالب مشترک تبدیل و رکوردهای قابل‌استفاده مشخص شوند.</p>
<p><strong>هدف این بخش:</strong> تبدیل داده خام به Dataset استاندارد و قابل‌تحلیل بدون حذف نامستند رکوردها.</p>
<p><strong>اصطلاحات کلیدی:</strong> <strong>Schema:</strong> ساختار استاندارد ستون‌ها؛ <strong>raw_harmonized:</strong> داده خام با ستون‌های هماهنگ‌شده؛ <strong>Eligibility:</strong> قواعد تعیین واجد شرایط بودن رکورد؛ <strong>Provenance:</strong> امکان ردیابی منشأ رکورد.</p>
<p>هر پلتفرم فرمت خام خودش رو داره، پس اول همه رو به یک Schema مشترک («raw_harmonized») نگاشت کردیم، بعد یک زنجیره‌ی قاعده‌مند از فیلترها («Eligibility») روش اجرا کردیم: اعتبارسنجی خام → حذف تکراری‌های دقیق → قاعده‌ی بازه‌ی زمانی → وجود متن → Provenance → ربط موضوعی → نقش تحلیلی. هیچ رکوردی بدون دلیل مستند حذف نمی‌شه — هرکدوم یه برچسب می‌خوره که چرا کجا رفته.</p>
</div>

```python
harmonized_counts = {}
for platform in ["x", "reddit", "youtube"]:
    d = ROOT / "data" / "raw_harmonized" / platform
    n = 0
    for f in d.glob("*.parquet"):
        try:
            n += len(pd.read_parquet(f, columns=["platform"]))
        except Exception:
            pass
    harmonized_counts[platform] = n

print("Harmonized raw records (raw_harmonized) by platform:")
for p, n in harmonized_counts.items():
    print(f"  {p:8s}: {n:>8,}")
print(f"  {'Total':8s}: {sum(harmonized_counts.values()):>8,}")
```

```python
audit = pd.read_parquet(ROOT / "data" / "audits" / "eligibility_audit.parquet")
funnel = audit["dataset_target"].value_counts()
eligible = funnel.get("opinion_main", 0) + funnel.get("opinion_limited", 0) + funnel.get("opinion_untimed", 0)

print(f"Total records entering Eligibility: {len(audit):,}\n")
print("Outcome of each record, by destination:")
print(funnel.to_string())
print(
    f"\nTotal Eligible records for opinion analysis "
    f"(opinion_main + opinion_limited + opinion_untimed): {eligible:,}"
)
```

<div dir="rtl" style="direction:rtl; text-align:right; font-family:Vazirmatn, Tahoma, Arial, sans-serif; line-height:1.9;">
<h3 style="color:#073B74;">۵. توافق بین دو Annotator</h3>
<p>اگر دو Annotator یک متن را متفاوت برچسب بزنند، باید بدانیم میزان این اختلاف چقدر است.</p>
<p><strong>هدف این بخش:</strong> سنجش میزان سازگاری برچسب‌گذاری انسانی.</p>
<p><strong>اصطلاحات کلیدی:</strong> <strong>Double Annotation:</strong> برچسب‌گذاری مستقل یک متن توسط دو نفر؛ <strong>Percent Agreement:</strong> درصد برچسب‌های یکسان؛ <strong>Cohen's Kappa:</strong> توافق با درنظرگرفتن توافق احتمالی ناشی از شانس.</p>
<p>روی همون ۱۲۰ رکورد double-annotated، توافق دو annotator رو با <strong>Cohen's Kappa</strong> سنجیدیم (نه فقط درصد توافق ساده، چون اون به‌تنهایی گمراه‌کننده‌ست — با شانس هم می‌شه درصد بالایی گرفت).</p>
</div>

```python
with open(ROOT / "outputs" / "audits" / "annotator_agreement.json", encoding="utf-8") as f:
    agreement = json.load(f)

rows = []

for axis, r in agreement.items():
    rows.append({
        "Axis": axis.replace("_label", ""),
        "n Compared": r["n_compared"],
        "Cohen's Kappa": round(r["cohens_kappa"], 3),
        "Agreement Rate": f'{r["percent_agreement"]:.1%}',
    })

pd.DataFrame(rows)
```

<h3 style="color:#073B74;">۶. انتخاب مدل برای Annotation</h3>
<p>قبل از اجرای Annotation در مقیاس بزرگ، چند مسیر مدل مقایسه می‌شوند تا یک گزینه ثابت برای ادامه پروژه انتخاب شود.</p>
<p><strong>هدف این بخش:</strong> انتخاب مدلی که بین کیفیت، Coverage، خطا و هزینه تعادل مناسبی داشته باشد.</p>
<p><strong>اصطلاحات کلیدی:</strong> <strong>Route:</strong> ترکیب Model و Provider؛ <strong>Macro-F1:</strong> میانگین F1 کلاس‌ها با وزن برابر؛ <strong>Coverage:</strong> سهم رکوردهایی که خروجی قابل‌استفاده دریافت می‌کنند؛ <strong>Failure Rate:</strong> سهم اجراهای ناموفق.</p>
<p>سه Route مدل/Provider مختلف (هرکدوم رایگان یا خیلی ارزون) رو دقیقاً روی همون ۳۰۰ رکورد Gold Sample تست کردیم و کنار هم گذاشتیم — نه فقط روی چند نمونه‌ی دستچین‌شده. معیار انتخاب: ترکیب Macro-F1، نرخ خطا، Coverage و هزینه، نه فقط ارزون‌ترین یا فقط دقیق‌ترین.</p>
</div>

```python
with open(ROOT / "outputs" / "model_evaluation" / "sentiment_accuracy_summary.json", encoding="utf-8") as f:
    model_eval = json.load(f)

rows = []
for route, r in model_eval["routes"].items():
    rows.append({
        "Route": route,
        "Coverage": f'{r["coverage"]:.1%}',
        "Failure Rate": f'{r["failure_rate"]:.1%}',
        "Cost per 1,000 Records ($)": round(r["cost_per_1000_records_usd"], 3),
        "Sentiment Macro-F1": round(r["sentiment"]["macro_f1"], 3),
        "Stance Macro-F1": round(r["stance"]["macro_f1"], 3),
    })
df_models = pd.DataFrame(rows).sort_values("Sentiment Macro-F1", ascending=False)
df_models
```

<h3 style="color:#073B74;">۷. Annotation کامل: مقیاس‌دادن بدون بودجه</h3>
<p>پس از انتخاب مدل، باید Annotation را روی حجم بزرگ‌تری از داده اجرا کنیم؛ محدودیت منابع تعیین می‌کند این اجرا چگونه برنامه‌ریزی شود.</p>

<p>برای اینکه از پس این محدودیت واقعی بربیایم:</p>

<p>برای هر هفته و هر پلتفرم، سهم هر کلاس Sentiment رو همراه با <strong>فاصله‌ی اطمینان ۹۵٪ Wilson</strong> گزارش کردیم (نه فقط یه عدد خام بدون incertitude) — هفته‌هایی با نمونه‌ی کم (<code style="direction:ltr; unicode-bidi:embed; font-family:ui-monospace,Consolas,'DejaVu Sans Mono',monospace; background:#47494b; border-radius:4px; padding:0.1em 0.4em; font-size:0.9em;">n&lt;30</code>) صریحاً علامت‌گذاری شدن، و W21 (پنج‌روزه و ناقص) هم جدا مشخصه.</p>

```python
weekly = pd.read_csv(ROOT / "outputs" / "tables" / "weekly_trend_sentiment_by_platform.csv", encoding="utf-8-sig")
weekly.head(12)
```

```python
import matplotlib.pyplot as plt

SENTIMENT_COLORS = {
    "positive": "#009E73",  # Green
    "negative": "#D55E00",  # Brick red
    "neutral":  "#999999",  # Gray
    "mixed":    "#CC79A7",  # Pink
}

platform_to_plot = "reddit"
subset = weekly[weekly["platform"] == platform_to_plot].copy()
subset = subset.sort_values("project_week")

fig, ax = plt.subplots(figsize=(9, 4.5))
for label, color in SENTIMENT_COLORS.items():
    line = subset[subset["class_label"] == label].sort_values("project_week")
    if line.empty:
        continue
    ax.plot(line["project_week"], line["class_proportion"], label=label, color=color, linewidth=2, marker="o", markersize=4)
    ax.fill_between(line["project_week"], line["wilson_ci_low"], line["wilson_ci_high"], color=color, alpha=0.12, linewidth=0)

ax.set_title(f"Weekly Sentiment Share — {platform_to_plot} (with 95% CI)")
ax.set_ylabel("Share of Total Annotated Records per Week")
ax.set_xlabel("Project Week")
ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="y", linewidth=0.5, alpha=0.3)
ax.tick_params(axis="x", rotation=60)
ax.legend(frameon=False, ncol=4, loc="upper center", bbox_to_anchor=(0.5, -0.28))
fig.tight_layout()
plt.show()
```

<p><strong>اصطلاحات کلیدی:</strong> <strong>Limitation:</strong> عاملی که دامنه یا دقت نتیجه را محدود می‌کند؛ <strong>Non-probability Sample:</strong> نمونه‌ای با احتمال انتخاب نامعلوم؛ <strong>Coverage:</strong> میزان پوشش واقعی داده یا مدل.</p>
<ul style="padding-inline-start:1.4em;"><li><strong>نمونه غیراحتمالیه.</strong> هیچ‌جای این نوت‌بوک از عباراتی مثل «افکار عمومی جهان» یا «همه‌ی کاربران» استفاده نشده و نباید بشه — فقط توصیف نمونه‌ی مشاهده‌شده‌ست.</li><li><strong>Annotation کامل هنوز در حال گسترشه.</strong> با ۱۱ کلید API رایگان/کم‌هزینه، الان بخش قابل‌توجهی از داده annotate شده (بخش ۷)؛ تکمیل annotation کل داده و افزودن Target‌های تکمیلی (<code style="direction:ltr; unicode-bidi:embed; font-family:ui-monospace,Consolas,'DejaVu Sans Mono',monospace; background:#47494b; border-radius:4px; padding:0.1em 0.4em; font-size:0.9em;">T04</code>-<code style="direction:ltr; unicode-bidi:embed; font-family:ui-monospace,Consolas,'DejaVu Sans Mono',monospace; background:#47494b; border-radius:4px; padding:0.1em 0.4em; font-size:0.9em;">T06</code>)، خارج از بازه‌ی این تحویل، ادامه پیدا می‌کنه — نه به این معنی که چیزی نادیده گرفته شده، بلکه یه تصمیم آگاهانه برای مدیریت منابع محدود بود که با نمونه‌گیری طبقه‌بندی‌شده (بخش ۷) جبران شد.</li><li><strong>کیفیت مدل کامل نیست.</strong> Macro-F1 مدل انتخاب‌شده روی Sentiment حدود ۰.۶۱ و روی Stance حدود ۰.۵۰ است (بخش ۶) — یعنی annotation خودکار خطا داره، نه یک مرجع قطعی.</li><li><strong>همبستگی، علیت نیست.</strong> هم رابطه با شاخص‌های مالی و هم اثر رویدادها، به‌عنوان «همراهی زمانی» گزارش شدن، نه رابطه‌ی علّی اثبات‌شده.</li><li><strong>Engagement فقط برای YouTube کامله</strong> — X و Reddit این فیلد رو پر نمی‌کنن، پس مقایسه‌های وزن‌دار به Engagement فقط برای YouTube معناداره.</li><li><strong>پوشش زبان نامتوازنه</strong> (فارسی/انگلیسی بیشتر از عربی) — دقیقاً طبق سهمیه‌ی از‌پیش‌تعیین‌شده‌ی Gold Sample (بخش ۴)، ولی همچنان یه محدودیت واقعیه که باید در تفسیر نتایج در نظر گرفته بشه.</li></ul>

<h3 style="color:#073B74;">جمع‌بندی</h3>
<p>این بخش مسیر پروژه را در چند جمله جمع می‌کند و نشان می‌دهد خروجی‌های اصلی چگونه به هم متصل هستند.</p>
<p><strong>هدف این بخش:</strong> مرور سریع کل پروژه و مشخص‌کردن قدم بعدی.</p>
<p><strong>اصطلاحات کلیدی:</strong> <strong>Decision Log:</strong> ثبت تصمیم‌ها و تغییرات روش؛ <strong>Claim Registry:</strong> ثبت ادعاهای نهایی و شواهد آن‌ها؛ <strong>Audit:</strong> بررسی قابلیت ردیابی و صحت اجرای مراحل.</p>
<p>از جمع‌آوری بیش از سیصد هزار رکورد خام از سه پلتفرم، تا Eligibility، Gold Sample، انتخاب و ارزیابی مدل annotation، و تحلیل‌های توصیفی/زمانی/رویدادی/مالی/حساسیت — همه‌ی مراحل این پروژه با مستندسازی کامل (Manifest، Decision Log، Claim Registry) پیش رفته، طوری که هر عدد قابل ردیابی به یک فایل مشخصه. ادامه‌ی طبیعی این کار، تکمیل annotation روی کل داده‌ی Eligible و گسترش تحلیل به Targetهای تکمیلیه.</p>
